```python
#41.Calculate the missing-value count and missing-value percentage for every column.
import pandas as pd
df=pd.read_csv("Online_Retail_Practice_Sample.csv")
missing_count = df.isna().sum()
missing_percentage = (df.isna().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

display(missing_summary)
```

```python
#42.Identify the columns with the highest percentage of missing values and rank them.
missing_rank = missing_summary.sort_values(
    by="Missing Percentage",
    ascending=False
)

display(missing_rank)
```

```python
#43.Display all records where CustomerID is missing and determine how many transactions are affected.
missing_customers = df[df["CustomerID"].isna()]

display(missing_customers)

print("Transactions with missing CustomerID:", len(missing_customers))
```

```python
#44.Display records where Description is missing or contains only whitespace.
missing_description = df[
    df["Description"].isna() |
    df["Description"].str.strip().eq("").fillna(False)
]

display(missing_description)

print("Records with missing or blank Description:", len(missing_description))
```

```python
#45.Create a missing-value report with column name, missing count, non-missing count and missing percentage.
missing_report = pd.DataFrame({
    "Column Name": df.columns,
    "Missing Count": df.isna().sum().values,
    "Non-Missing Count": df.notna().sum().values,
    "Missing Percentage": (df.isna().sum().values / len(df)) * 100
})

display(missing_report)
```

```python
#46.Identify exact duplicate rows and report the duplicate count before removing anything.
duplicate_count = df.duplicated().sum()

print("Exact duplicate rows:", duplicate_count)

display(df[df.duplicated(keep=False)])
```

```python
#47.Remove exact duplicate rows from a working copy and verify how many records remain.
df_clean = df.copy()

rows_before = len(df_clean)

df_clean = df_clean.drop_duplicates()

rows_after = len(df_clean)

print("Rows before removal:", rows_before)
print("Rows after removal:", rows_after)
print("Duplicates removed:", rows_before - rows_after)
print("Remaining duplicate rows:", df_clean.duplicated().sum())
```

```python
#48.Investigate repeated InvoiceNo and StockCode combinations without automatically deleting them, since multiple lines may be legitimate.
repeated_combinations = df_clean[
    df_clean.duplicated(
        subset=["InvoiceNo", "StockCode"],
        keep=False
    )
].sort_values(["InvoiceNo", "StockCode"])

display(repeated_combinations)

print("Records in repeated combinations:", len(repeated_combinations))
```

```python
#49.Handle missing Description values using a documented strategy and explain why missing CustomerID values should not be filled with invented IDs.
df_clean["Description"] = df_clean["Description"].replace(
    r"^\s*$",
    pd.NA,
    regex=True
)

df_clean["Description"] = df_clean["Description"].fillna(
    "Unknown Product"
)

print("Remaining missing descriptions:", df_clean["Description"].isna().sum())
```

```python
#50.Create a data-quality report comparing row counts, missing values and duplicate counts before and after cleaning.
df_before = df.copy()

df_after = df_clean.copy()

quality_report = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Missing Values",
        "Exact Duplicate Rows"
    ],
    "Before Cleaning": [
        len(df_before),
        len(df_before.columns),
        df_before.isna().sum().sum(),
        df_before.duplicated().sum()
    ],
    "After Cleaning": [
        len(df_after),
        len(df_after.columns),
        df_after.isna().sum().sum(),
        df_after.duplicated().sum()
    ]
})

display(quality_report)
```
