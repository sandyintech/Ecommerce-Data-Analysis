```python
#31.	Create a working copy of the original DataFrame and retain the untouched source as df_raw.
import pandas as pd
df = pd.read_csv("Online_Retail_Practice_Sample.csv")
df_raw = df.copy()
df = df_raw.copy()
print("Original dataset:")
display(df_raw.head())
print("Working dataset:")
display(df.head())
print("Original shape:", df_raw.shape)
print("Working shape:", df.shape)
```

```python
#32.	Rename the Description column to ProductDescription in your working copy.
df = df.rename(columns={
    "Description": "ProductDescription"
})
print("Updated column names:")
print(df.columns.tolist())
display(df.head())
```

```python
#33.	Create a Revenue column by multiplying Quantity by UnitPrice, preserving the original signed quantity.
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
display(df[[
    "InvoiceNo",
    "ProductDescription",
    "Quantity",
    "UnitPrice",
    "Revenue"
]].head(10))
```

```python
#34.	Create a TransactionType column to classify records as cancellation, return, positive sale or other/invalid, using documented rules.
import numpy as np
conditions = [
    df["InvoiceNo"].isna(),
    df["InvoiceNo"].astype("string").str.startswith(
        "C", na=False
    ),
    df["Quantity"].isna() | (df["Quantity"] == 0),
    df["Quantity"] < 0,
    df["Quantity"] > 0
]
categories = [
    "Other/invalid",
    "Cancellation",
    "Other/invalid",
    "Return",
    "Positive sale"
]
df["TransactionType"] = np.select(
    conditions,
    categories,
    default="Other/invalid"
)
display(df[[
    "InvoiceNo",
    "ProductDescription",
    "Quantity",
    "UnitPrice",
    "Revenue",
    "TransactionType"
]].head(15))
```

```python
#35.	Create an IsIdentifiedCustomer column that indicates whether CustomerID is available.
df["IsIdentifiedCustomer"] = df["CustomerID"].notna()
display(df[[
    "CustomerID",
    "IsIdentifiedCustomer"
]].head(15))
```

```python
#36.	Convert InvoiceDate to a proper Pandas datetime data type.
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)
print("InvoiceDate data type:")
print(df["InvoiceDate"].dtype)
display(df[["InvoiceNo", "InvoiceDate"]].head(10))
```

```python
#37.	Convert InvoiceNo, StockCode and Country to suitable string data types.
string_columns = [
    "InvoiceNo",
    "StockCode",
    "Country"
]
for column in string_columns:
    df[column] = df[column].astype("string")
print("Updated data types:")
print(df[string_columns].dtypes)
```

```python
#38.	Reorder columns so transaction identifiers and dates appear before product, customer and financial information.
priority_columns = [
    "InvoiceNo",
    "InvoiceDate",
    "StockCode",
    "ProductDescription",
    "CustomerID",
    "Country",
    "Quantity",
    "UnitPrice",
    "Revenue",
    "TransactionType",
    "IsIdentifiedCustomer"
]
remaining_columns = [
    column for column in df.columns
    if column not in priority_columns
]
df = df[priority_columns + remaining_columns]
print("Reordered columns:")
print(df.columns.tolist())
display(df.head())
```

```python
#39.	Sort the working dataset by InvoiceDate and then by InvoiceNo.
df = df.sort_values(
    by=["InvoiceDate", "InvoiceNo"],
    ascending=[True, True],
    na_position="last"
)
display(df[[
    "InvoiceDate",
    "InvoiceNo",
    "ProductDescription",
    "Quantity",
    "UnitPrice",
    "Revenue"
]].head(20))
```

```python
#40.	Reset the DataFrame index after sorting and verify that the new index is sequential.
df = df.reset_index(drop=True)
print("First 10 index values:")
print(df.index[:10].tolist())
print("Is the index sequential?")
print(df.index.equals(pd.RangeIndex(start=0, stop=len(df))))
print("Final dataset shape:", df.shape)
display(df.head(10))
```
