```python
#21.	Select and display only InvoiceNo and InvoiceDate for all transactions.
import pandas as pd
df=pd.read_csv("online_retail_practice_sample.csv")
df[['InvoiceNo', 'InvoiceDate']]
```

```python
#22.	Retrieve a transaction at a specified row position using iloc and explain the difference between row position and index label.
transaction = df.iloc[5]
print(transaction)
```

```python
#23.	Use loc to retrieve selected columns for a chosen range of index labels.
result = df.loc[0:4, ["InvoiceNo", "Description", "Quantity", "UnitPrice"]]
print(result)
```

```python
df.loc[5:10, ["InvoiceNo", "StockCode", "Country"]]
```

```python
#24.	Filter transactions where Quantity is greater than 10.
filtered_df = df[df["Quantity"] > 10]
print(filtered_df)
```

```python
#25.	Find transactions where UnitPrice is greater than £50.
filtered_df = df[df["UnitPrice"] > 50]
print(filtered_df)
```

```python
#26.	Retrieve all transactions belonging to a specified country using a Boolean condition.
uk_transactions = df[df["Country"] == "United Kingdom"]
print(uk_transactions)
```

```python
country_transactions = df[df["Country"] == "Germany"]
print(country_transactions)
```

```python
#27.	Use isin to find transactions from the five countries with the highest transaction counts.
top_5_countries = df["Country"].value_counts().head(5).index
filtered_df = df[df["Country"].isin(top_5_countries)]
print(filtered_df)
```

```python
#28.	Use between to retrieve transactions with quantities between 5 and 20 inclusive.
filtered_df = df[df["Quantity"].between(5, 20)]
print(filtered_df)
print("Total matching transactions:", len(filtered_df))
```

```python
#29.	Use query to find transactions where Quantity is greater than 10 and UnitPrice is greater than £5.

filtered_df = df.query("Quantity > 10 and UnitPrice > 5")
print("Transactions with Quantity > 10 and UnitPrice > 5:")
display(filtered_df)
print("Total matching transactions:", len(filtered_df))
```

```python
#30.	Sort transactions by UnitPrice in descending order and display the 20 highest unit-price records.
sorted_df = df.sort_values(by="UnitPrice",ascending=False)
top_20 = sorted_df.head(20)

print("20 transactions with the highest unit prices:")
display(top_20)

print("Number of records displayed:", len(top_20))
```
